```python
class Fraction:
    
    def __init__(self,x,y):
        self.num=x
        self.den=y
    
    def __str__(self):
        return '{}/{}'.format(self.num,self.den)
    
    def __add__(self,other):
        new_num=self.num*other.den +other.num*self.den
        new_den=self.den*other.den

        return '{}/{}'.format(new_num,new_den)
        print('idhar')
    
    def __sub__(self,other):
        new_num=self.num*other.den - other.num*self.den
        new_den=self.den*other.den

        return '{}/{}'.format(new_num,new_den)
        print('idhar')
    
    def __mul__(self,other):
        new_num=self.num*other.num
        new_den=self.den*other.den

        return '{}/{}'.format(new_num,new_den)
        print('idhar')
   
    def __truediv__(self,other):
        new_num=self.num*other.den
        new_den=self.den*other.num

        return '{}/{}'.format(new_num,new_den)
        print('idhar')
   
    def converttodecimal(self):
        return self.num/self.den
        
    
    
```

```python
fr1=Fraction(3,4)
fr2=Fraction(1,2)
```

```python
print(fr1+fr2)
print(fr1-fr2)
print(fr1*fr2)
print(fr1/fr2)
```

```python
fr1.converttodecimal()
```

```python
fr2.converttodecimal()
```

```python
class Point:
    def __init__(self,x,y):
        self.x_coord=x
        self.y_coord=y
        #<x,y>
    def __str__(self):
        return '<{},{}>'.format(self.x_coord,self.y_coord)
    def euclidean_dist(self,other):#self(x1,y1) other(x2,y2)
        #sqrt((x2-x1)**2+(y2-y1)**2)
        return ((self.x_coord-other.x_coord)**2 +(self.y_coord-other.y_coord)**2)**0.5

    def dist_from_origin(self):
        return self.euclidean_dist(Point(0,0))
```

```python
p1=Point(1,2)
p2=Point(2,3)
```

```python
print(p1)
```

```python
p1.euclidean_dist(p2)
```

```python
p1.dist_from_origin()
```

```python
class Line:
    def __init__(self,a,b,c):
        self.a=a
        self.b=b
        self.c=c
    def __str__(self):
        return '{}x+{}y+{} =0 '.format(self.a,self.b,self.c)
    def point_on_line(line,point):
        if line.a*point.x_coord+line.b*point.y_coord+line.c==0:
            return 'lies on the line'
        else:
            return 'does not lie on the line'
    def short_dist(line,point):
        return abs(line.a*point.x_coord+line.b*point.y_coord+line.c)/(line.a**2+line.b**2)**0.5
```

```python
l1=Line(1,1,-2)
p1=Point(0,1)
```

```python
print(l1)
```

```python
l1.point_on_line(p1)
```

```python
l1.short_dist(p1)
```

```python
class Phone:
    def __init__(self,price,brand,camera):
        print('Inside Phone Constructor')
        self.__price=price
        self.brand=brand
        self.camera=camera
class SmartPhone(Phone):#triggering Parent constructor and initializing via Super()
    def __init__(self,price,brand,camera,os,ram):
        super().__init__(price,brand,camera)
        self.os=os
        self.ram=ram
        print('Inside smartphone Constructor')

t=SmartPhone(20000,'Samsung',12,'Android',2)

print(t.os)
print(t.brand)
print(t.camera)
```

```python
# the diamond problem
# https://stackoverflow.com/questions/56361048/what-is-the-diamond-problem-in-python-and-why-its-not-appear-in-python2
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print ("Buying a phone")

class Product:
    def buy(self):
        print ("Product buy method")

# Method resolution order
class SmartPhone(Phone,Product):
    pass

s=SmartPhone(20000, "Apple", 12)

s.buy()
```

```python
#Abstraction
#Applying Contraints on Lower Classes
#It has atleast 1 abstract method(Without Code)
from abc import ABC,abstractmethod
class BankApp(ABC):
    def dbase(self):
        print('Connected with database')
    @abstractmethod
    def security(self):
        pass
    @abstractmethod
    def display(self):
        pass
```

```python
type(BankApp)
```

```python
class MobileApp(BankApp):
    def mobile_login(self):
        print('Print into mobile')
    def security(self):
        print('mobile security')
    def display(self):
        print('Mobile Interface')
```

```python
mob=MobileApp()
print(mob)
```

```python
mob.security()
```

```python
mob.display()
```
