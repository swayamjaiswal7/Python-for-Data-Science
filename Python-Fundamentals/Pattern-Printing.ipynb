- 1 Between 1 to 100 print all numbers multiple of 4 using while loop

```python
i=1
while i<=100:
    if i%4==0:
        print(i,end=",")
    i+=1
```

```python
n=5
for i in range(n):
    for j in range(2*n):
        if j<n-i :
            print("*",end="")
        elif j>n-i-1 and j<n+i:
            print(" ",end="")
        else:
            print("*",end="")
    print()
```

```python
N = 5

for i in range(2 * N):          
    for j in range(2 * N):      
        if i < N:
            if j < N - i or j >= N + i:
                print("*", end="")
            else:
                print(" ", end="")
        else:
            if j < i - N + 1 or j >= 3 * N - 1 - i:
                print("*", end="")
            else:
                print(" ", end="")
    print()
```

```python
N = 5  # Number of rows

for i in range(N):
    # Print leading spaces
    for _ in range(N - i - 1):
        print("_", end="")

    for j in range(2 * i + 1):
        if j % 2 == 0:
            print("*", end="")
        else:
            print("_", end="")

    print()  
```
