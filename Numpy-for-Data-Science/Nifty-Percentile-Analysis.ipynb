```python
#Extracting Dataset
!pip install yfinance
import pandas as pd
import yfinance as yf

nifty = yf.download("^NSEI", period="5y", interval="1d")
```

```python
# Dropping Null values
nifty = nifty.dropna()
```

```python
#Extracting Month from Date
nifty['YearMonth'] = nifty.index.to_period('M')
```

```python
#Function for monthly percentiles
def calculate_monthly_percentiles(df):
    results = {}

    for period, group in df.groupby('YearMonth'):
        prices = group['Close'].to_numpy()
        n = len(prices)

        # Store percentiles for each starting day
        percentile_matrix = [np.percentile(prices[start:] - prices[start], range(10, 101, 10)) for start in range(n)]

        results[period] = np.array(percentile_matrix)

    return results
```

```python
monthly_percentiles = calculate_monthly_percentiles(nifty)
```

```python
latest_month = list(monthly_percentiles.keys())[-1]
```

```python
matrix = monthly_percentiles[latest_month]
```
