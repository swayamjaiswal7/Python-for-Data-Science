```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

def get_naukri_jobs(query="data analyst", location="India", pages=2):
    base_url = "https://www.naukri.com/"
    query = query.replace(" ", "-")
    jobs_data = []

    for page in range(pages):
        url = f"{base_url}{query}-jobs-{page+1}?k={query}&l={location}"
        headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/135.0.0.0 Safari/537.36"}

        print(f"Scraping Page {page+1}...")

        res = requests.get(url, headers=headers)
        if res.status_code != 200:
            print("Failed to retrieve page:", url)
            continue

        soup = BeautifulSoup(res.text, 'html.parser')
        job_cards = soup.find_all('article', class_='jobTuple')

        for card in job_cards:
            try:
                title = card.find('a', class_='title').text.strip()
                company = card.find('a', class_='subTitle').text.strip()
                location = card.find('li', class_='location').text.strip()
                exp = card.find('li', class_='exp').text.strip()
                description = card.find('ul', class_='tags').text.strip()
                link = card.find('a', class_='title')['href']

                jobs_data.append({
                    'Title': title,
                    'Company': company,
                    'Location': location,
                    'Experience': exp,
                    'Short Description': description,
                    'Link': link
                })
            except AttributeError:
                continue

        time.sleep(1)  # polite crawling

    return pd.DataFrame(jobs_data)

# Run the scraper
df = get_naukri_jobs(query="data analyst", location="Mumbai", pages=3)

# Save to CSV
df.to_csv("naukri_jobs_data.csv", index=False)
print("Saved to naukri_jobs_data.csv")
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
import time
import pandas as pd

def naukri_job_scraper(search_term="data analyst", location="mumbai", pages=2):
    options = Options()
    options.add_argument("--headless")  # Run without opening browser
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")

    driver = webdriver.Chrome(options=options)

    base_url = "https://www.naukri.com/{}-jobs?k={}&l={}&nignbevent_src=jobsearchDeskGNB"
    query = search_term.replace(" ", "-")

    jobs = []

    for page in range(1, pages+1):
        url = base_url.format(query, search_term, location) + f"&pageNo={page}"
        print(f"Loading page {page}: {url}")
        driver.get(url)
        time.sleep(3)  # Let JS load

        job_cards = driver.find_elements(By.CLASS_NAME, "jobTuple")

        for job in job_cards:
            try:
                title = job.find_element(By.CLASS_NAME, "title").text
                company = job.find_element(By.CLASS_NAME, "subTitle").text
                location = job.find_element(By.CLASS_NAME, "location").text
                experience = job.find_element(By.CLASS_NAME, "exp").text
                job_link = job.find_element(By.CLASS_NAME, "title").get_attribute("href")

                try:
                    tags = job.find_elSement(By.CLASS_NAME, "tags").text
                except:
                    tags = ""

                jobs.append({
                    "Title": title,
                    "Company": company,
                    "Location": location,
                    "Experience": experience,
                    "Tags": tags,
                    "Job URL": job_link
                })
            except Exception as e:
                print("Error parsing a job card:", e)

        time.sleep(2)

    driver.quit()

    df = pd.DataFrame(jobs)
    df.to_csv("naukri_jobs_data.csv", index=False)
    print("Saved to naukri_jobs_data.csv")
    return df

# Run it
df = naukri_job_scraper("data", "mumbai", pages=3)
print(df.head())
```

```python
kohli_df = ipl[ipl['batsman'] == 'V Kohli']

# === Part 1: Highest Score Against Each Team ===

# Total runs per match against each team
match_scores = kohli_df.groupby(['match_id', 'bowling_team'])['batsman_runs'].sum().reset_index()

# Max score per team
max_scores = match_scores.groupby('bowling_team')['batsman_runs'].max().reset_index()

# Add corresponding match_id
highest_scores = pd.merge(match_scores, max_scores,
                          on=['bowling_team', 'batsman_runs'], how='inner')
highest_scores = highest_scores.rename(columns={
    'bowling_team': 'Opponent Team',
    'batsman_runs': 'Highest Score',
    'match_id': 'Match ID'
})

# === Part 2: Average Against Each Team ===

# Total runs per team
total_runs = kohli_df.groupby('bowling_team')['batsman_runs'].sum().reset_index(name='Total Runs')

# Number of times dismissed per team
dismissals = kohli_df[kohli_df['player_dismissed'] == 'V Kohli']
dismissal_counts = dismissals.groupby('bowling_team').size().reset_index(name='Dismissals')

# Merge and calculate average
avg_df = pd.merge(total_runs, dismissal_counts, on='bowling_team', how='left')
avg_df['Dismissals'] = avg_df['Dismissals'].fillna(0)
avg_df['Average'] = avg_df.apply(
    lambda row: row['Total Runs'] / row['Dismissals'] if row['Dismissals'] > 0 else float('inf'),
    axis=1
)

# === Final Merge ===

# Merge averages with highest score info
final_df = pd.merge(highest_scores, avg_df, left_on='Opponent Team', right_on='bowling_team', how='left')
final_df.drop(columns=['bowling_team'], inplace=True)

# Optional: Round off values
final_df['Average'] = final_df['Average'].round(2)

# Final output
print(final_df[['Opponent Team', 'Highest Score', 'Match ID', 'Total Runs', 'Dismissals', 'Average']])
```

```python
kohli_df = ipl[ipl['batsman'] == 'V Kohli']

# === Highest Score Per Team ===
match_scores = kohli_df.groupby(['match_id', 'bowling_team'])['batsman_runs'].sum().reset_index()
max_scores = match_scores.groupby('bowling_team')['batsman_runs'].max().reset_index()
highest_scores = pd.merge(match_scores, max_scores,
                          on=['bowling_team', 'batsman_runs'], how='inner')
highest_scores = highest_scores.rename(columns={
    'bowling_team': 'Opponent Team',
    'batsman_runs': 'Highest Score',
    'match_id': 'Match ID'
})

# === Average Per Team ===
total_runs = kohli_df.groupby('bowling_team')['batsman_runs'].sum().reset_index(name='Total Runs')
dismissals = kohli_df[kohli_df['player_dismissed'] == 'V Kohli']
dismissal_counts = dismissals.groupby('bowling_team').size().reset_index(name='Dismissals')
avg_df = pd.merge(total_runs, dismissal_counts, on='bowling_team', how='left')
avg_df['Dismissals'] = avg_df['Dismissals'].fillna(0)
avg_df['Average'] = avg_df.apply(
    lambda row: row['Total Runs'] / row['Dismissals'] if row['Dismissals'] > 0 else float('inf'),
    axis=1
)

# === Strike Rate Per Team ===
balls_faced = kohli_df[~kohli_df['extra_runs'].isin([1])]  # Filters out extras that don't count as balls faced (optionally improve)
balls_per_team = kohli_df.groupby('bowling_team').size().reset_index(name='Balls Faced')

# Merge with average and total
avg_df = pd.merge(avg_df, balls_per_team, on='bowling_team', how='left')
avg_df['Strike Rate'] = (avg_df['Total Runs'] / avg_df['Balls Faced']) * 100
avg_df['Strike Rate'] = avg_df['Strike Rate'].round(2)

# === Final Merge ===
final_df = pd.merge(highest_scores, avg_df, left_on='Opponent Team', right_on='bowling_team', how='left')
final_df.drop(columns=['bowling_team'], inplace=True)
final_df['Average'] = final_df['Average'].round(2)

# Final output
print(final_df[['Opponent Team', 'Highest Score', 'Match ID', 'Total Runs', 'Dismissals', 'Average', 'Balls Faced', 'Strike Rate']])
```

```python
import pandas as pd

class BatsmanStats:
    def __init__(self, df: pd.DataFrame, batsman_name: str):
        self.df = df
        self.batsman_name = batsman_name
        self.batsman_df = self.df[self.df['batsman'] == batsman_name]
    
    def highest_score_against_each_team(self):
        match_scores = self.batsman_df.groupby(['match_id', 'bowling_team'])['batsman_runs'].sum().reset_index()
        max_scores = match_scores.groupby('bowling_team')['batsman_runs'].max().reset_index()
        highest = pd.merge(match_scores, max_scores, on=['bowling_team', 'batsman_runs'], how='inner')
        highest = highest.rename(columns={
            'bowling_team': 'Opponent Team',
            'batsman_runs': 'Highest Score',
            'match_id': 'Match ID'
        })
        return highest

    def average_and_strike_rate(self):
        # Total runs
        total_runs = self.batsman_df.groupby('bowling_team')['batsman_runs'].sum().reset_index(name='Total Runs')

        # Dismissals
        dismissals_df = self.batsman_df[self.batsman_df['player_dismissed'] == self.batsman_name]
        dismissals = dismissals_df.groupby('bowling_team').size().reset_index(name='Dismissals')

        # Balls faced (exclude wides)
        legal_balls = self.batsman_df[~self.batsman_df['extra_runs'].isin([1])]  # Optionally filter more accurately
        balls_faced = legal_balls.groupby('bowling_team').size().reset_index(name='Balls Faced')

        # Merge all
        stats = pd.merge(total_runs, dismissals, on='bowling_team', how='left')
        stats['Dismissals'] = stats['Dismissals'].fillna(0)
        stats = pd.merge(stats, balls_faced, on='bowling_team', how='left')

        # Average and Strike Rate
        stats['Average'] = stats.apply(
            lambda row: row['Total Runs'] / row['Dismissals'] if row['Dismissals'] > 0 else float('inf'),
            axis=1
        )
        stats['Strike Rate'] = (stats['Total Runs'] / stats['Balls Faced']) * 100

        stats = stats.rename(columns={'bowling_team': 'Opponent Team'})
        stats['Average'] = stats['Average'].round(2)
        stats['Strike Rate'] = stats['Strike Rate'].round(2)

        return stats

    def complete_summary(self):
        high_scores = self.highest_score_against_each_team()
        avg_sr = self.average_and_strike_rate()
        summary = pd.merge(high_scores, avg_sr, on='Opponent Team', how='left')
        return summary[['Opponent Team', 'Highest Score', 'Match ID', 'Total Runs', 'Dismissals', 'Average', 'Balls Faced', 'Strike Rate']]
```

```python
msd_stats = BatsmanStats(ipl, batsman_name="MS Dhoni")
summary_df = msd_stats.complete_summary()
```
