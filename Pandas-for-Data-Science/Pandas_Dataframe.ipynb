```python
#Head to Head Record generator
def head_2_head(team1,team2):
    m1=(ipl.Team1==team1)|(ipl.Team2==team1)
    m2=(ipl.Team1==team2)|(ipl.Team2==team2)
    df1=ipl[m1&m2]
    print(df1.WinningTeam.value_counts())
    print(df1.Player_of_Match.value_counts().head(10))
```

```python
head_2_head('Chennai Super Kings','Delhi Capitals')
```
