```python
tips=sns.load_dataset('tips')
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import numpy as np
import seaborn as sns
```

```python
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex',style='time',size='size')
plt.show()
```

```python
#figure level function
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex',style='time',size='size')
```

```python
gap = px.data.gapminder()
temp_df = gap[gap['country'] == 'India']
temp_df
```

```python
sns.lineplot(data=temp_df, x='year', y='lifeExp')
```

```python
sns.lineplot(data=temp_df,x='year',y='pop')
```

```python
temp_df = gap[gap['country'].isin(['India','Brazil','Germany'])]
temp_df
```

```python
sns.relplot(kind='line',data=temp_df,x='year',y='lifeExp',hue='country',style='continent',size='continent')
```

```python
#not used in access function
#only with figure level function(rel)
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', col='sex', row='day')
```

```python
# col wrap
sns.relplot(data=gap, x='lifeExp', y='gdpPercap', kind='scatter', col='year', col_wrap=3)
```

## Distribution Plots
- used for univariate analysis
- used to find out the distribution
- Range of the observation
- Central Tendency
- is the data bimodal?
- Are there outliers?

Plots under distribution plot

- histplot
- kdeplot
- rugplot

```python
#univariate analysis
sns.histplot(data=tips,x='total_bill')
```

```python
sns.displot(data=tips, x='total_bill', kind='kde')
```

```python
sns.displot(data=tips, x='total_bill', kind='hist',bins=2)
```

```python
# It’s also possible to visualize the distribution of a categorical variable using the logic of a histogram. 
# Discrete bins are automatically set for categorical variables

# countplot
sns.displot(data=tips, x='day', kind='hist',hue='sex') 
```

```python
sns.displot(data=tips,x='time',kind='hist')
```

```python
sns.displot(data=tips,x='tip',hue='sex')
```

```python
sns.displot(data=tips,x='tip',hue='sex',element='step')
```

```python
sns.displot(data=tips,x='total_bill')
```

```python
sns.displot(data=tips,x='size')
```

```python
titanic = sns.load_dataset('titanic')
titanic
```

```python
sns.displot(data=titanic, x='age', kind='hist',element='step',hue='sex')
```

```python
# faceting using col and row -> not work on histplot function
sns.displot(data=tips, x='tip', kind='hist',col='sex',element='step')
```

```python
sns.kdeplot(data=tips,x='total_bill')
```

```python
sns.kdeplot(data=tips,x='total_bill',hue='sex',fill=True)
```

```python
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex',fill=True,height=10,aspect=2)
```

```python
#Rug Plot
#Marginal distribution by plotting ticks along x and y
sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill')
```

```python
#Bivariate Analysis using Histogram

sns.histplot(data=tips,x='total_bill',y='tip')
```

```python
sns.displot(data=tips,x='total_bill',y='tip')
```

```python
sns.kdeplot(data=tips,x='total_bill',y='tip')
```

## Matrix plot

```python
gap
```

```python
heat=gap.pivot(index='country',columns='year',values='lifeExp')
```

```python
sns.heatmap(heat)
plt.figure(figsize=(55,10))
```

```python
europe=gap[gap['continent']=='Europe']
```

```python
europe=europe.pivot(index='country',columns='year',values='lifeExp')
```

```python
sns.heatmap(europe,annot=True,linewidths=0.5,cmap='jet')
plt.figure(figsize=(20,10))
```
