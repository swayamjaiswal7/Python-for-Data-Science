```python
df=pd.read_csv('batsman_season_record.csv')
```

```python
df
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Basic data
years = ['2015', '2016', '2017']
x = np.arange(len(df['batsman']))  # numeric positions for batsmen
width = 0.25  # width of each bar

# Plot bars for each year
plt.bar(x - width, df['2015'], width=width, label='2015')
plt.bar(x,         df['2016'], width=width, label='2016')
plt.bar(x + width, df['2017'], width=width, label='2017')

# Add labels and formatting
plt.xticks(x, df['batsman'], rotation=45)
plt.xlabel('Batsman')
plt.ylabel('Runs')
plt.title('Runs Scored by Batsman (2015-2017)')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.bar(np.arange(df.shape[0]) - 0.2,df['2015'],width=0.2,color='yellow')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2,color='red')
plt.bar(np.arange(df.shape[0]) + 0.2,df['2017'],width=0.2,color='blue')

plt.xticks(np.arange(df.shape[0]), df['batsman'])

plt.show()
```

```python
#Stacked Bar Chart
plt.bar(df['batsman'],df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016')
plt.bar(df['batsman'],df['2015'],bottom=(df['2016'] + df['2017']),label='2015')

plt.legend()
plt.show()
```

```python
batters=pd.read_csv('batter_ipl.csv')
```

```python
#Subplots
batters.head(10)
```

```python
fig,ax=plt.subplots()
ax.scatter(batters['avg'],batters['strike_rate'],color='green')
ax.set_title('Avg vs Strike Rate')
ax.set_xlabel('Average')
ax.set_ylabel('Strike Rate')
fig.show()
```

```python
fig,ax=plt.subplots(nrows=2,ncols=1,figsize=(10,6),sharex=True)
ax[0].scatter(batters['avg'],batters['strike_rate'],color='green')
ax[0].set_title('Avg Vs Strike Rate')
ax[1].scatter(batters['avg'],batters['runs'])
ax[0].set_ylabel('Strike Rate')

ax[1].set_title('Avg Vs Runs')
ax[1].set_ylabel('Runs')
ax[1].set_xlabel('Avg')
```

```python
fig, ax = plt.subplots(nrows=2,ncols=2,figsize=(10,10))

ax[0,0].scatter(batters['avg'],batters['strike_rate'])
ax[0,1].scatter(batters['avg'],batters['runs'])
ax[1,0].hist(batters['avg'])
ax[1,1].hist(batters['runs'])
```

```python
fig=plt.figure()
ax1=fig.add_subplot(3,1,1)
ax1.scatter(batters['avg'],batters['strike_rate'],color='yellow')
ax2=fig.add_subplot(3,1,2)
ax2.hist(batters['runs'])
ax3=fig.add_subplot(3,1,3)
ax3.bar(batters['batter'].head(10),batters['runs'].head(10),color='skyblue')
ax3.tick_params('x',rotation=50)
ax3.bar_label(ax3.bar(batters['batter'].head(10),batters['runs'].head(10),color='skyblue')
,labels=batters['runs'].head())
fig.show()
```

```python
#3d surface plot
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
```

```python
xx, yy = np.meshgrid(x,y)
```

```python
z=xx**2+yy**2
z.shape
```

```python
fig=plt.figure(figsize=(12,7))
ax=plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
z=np.sin(xx)+np.cos(yy)
```

```python
fig=plt.figure(figsize=(12,7))
ax=plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
#scatter+surface
z=xx**2+yy**2
fig=plt.figure(figsize=(12,7))
ax=plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
ax=plt.subplot()
p=ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
fig=plt.figure(figsize=(12,7))
ax=plt.subplot()
p=ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```
